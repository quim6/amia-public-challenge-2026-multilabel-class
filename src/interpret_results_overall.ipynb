```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load data
df = pd.read_csv("models\per_class_training_metrics_model_efficientnet_b4_contrast.csv")

# Set up figure
plt.figure(figsize=(10, 6))

# Line chart grouping by class
sns.lineplot(data=df, x="epoch", y="roc_auc", hue="class", marker="o")

plt.title("Per-Class ROC-AUC Across Epochs - CLAHE", fontsize=14, fontweight="bold")
plt.xlabel("Epoch", fontsize=12)
plt.ylabel("ROC-AUC", fontsize=12)
plt.legend(title="Class", bbox_to_anchor=(1.05, 1), loc="upper left")
plt.grid(True, linestyle="--", alpha=0.6)
plt.tight_layout()
plt.savefig("figs/per_class_roc_contrast.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load overall training metrics file
df = pd.read_csv("models/training_metrics_model_efficientnet_b4.csv")

# Setup two side-by-side plots
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Plot 1: Loss across epochs
axes[0].plot(df["epoch"], df["train_loss"], label="Train Loss", color="navy", linestyle="--", marker="o")
axes[0].plot(df["epoch"], df["val_loss"], label="Val Loss", color="crimson", linewidth=2, marker="s")
axes[0].set_title("Training & Validation Loss - no preprocessing", fontsize=12, fontweight="bold")
axes[0].set_xlabel("Epoch")
axes[0].set_ylabel("Loss")
axes[0].grid(True, linestyle="--", alpha=0.6)
axes[0].legend()

# Plot 2: ROC-AUC across epochs
axes[1].plot(df["epoch"], df["macro_roc_auc"], label=f"Macro ROC-AUC {df['macro_roc_auc'].max().round(3)}", color="navy", linestyle="--", marker="o")
axes[1].plot(df["epoch"], df["macro_pr_auc"], label=f"Macro PR-AUC {df['macro_pr_auc'].max().round(3)}", color="forestgreen", linewidth=2, marker="s")
axes[1].set_title("AUC metrics - no preprocessing", fontsize=12, fontweight="bold")
axes[1].set_xlabel("Epoch")
axes[1].set_ylabel("Mean ROC-AUC")
axes[1].grid(True, linestyle="--", alpha=0.6)
axes[1].legend()

plt.tight_layout()
plt.savefig("figs/loss_basic.png")
plt.show()
```
